## Schrieffer–Wolff (SW) transformation

The **Schrieffer–Wolff (SW) transformation** is a very useful tool in quantum mechanics and quantum information. It’s essentially a **unitary transformation** that allows you to derive an **effective Hamiltonian** when two subsystems are coupled but have a large energy detuning.

---

### 🔹 Intuition

Suppose you have a Hamiltonian split into two parts:

$$
H = H_0 + V
$$

* $H_0$: The "unperturbed" part, with well-separated energy levels (e.g., qubit energies).
* $V$: A weak interaction that couples these levels.

If some states are far detuned in energy (off-resonant), then transitions to them are strongly suppressed. But the interaction $V$ still produces **virtual effects** on the low-energy subspace.

The **SW transformation** is a way to **eliminate those off-resonant couplings** and derive an **effective Hamiltonian** that only acts within the low-energy subspace, but still accounts for the virtual processes.

---

### 🔹 How it works

1. You look for a unitary transformation

$$
\tilde{H} = e^S H e^{-S}
$$

where $S$ is an **anti-Hermitian operator** ($S^\dagger = -S$).

2. $S$ is chosen such that in the transformed Hamiltonian, the off-diagonal couplings between low-energy and high-energy states are canceled to leading order.

3. The result is an **effective Hamiltonian**:

$$
H_{\text{eff}} = H_0 + \frac{1}{2}[S, V] + \dots
$$

which acts only on the "relevant" subspace, with corrections coming from virtual excitations.

---

### 🔹 Example: Qubit–cavity system

In circuit QED, you have

$$
H = \omega_c a^\dagger a + \frac{\omega_q}{2} \sigma_z + g(a^\dagger \sigma_- + a \sigma_+)
$$

where:

* $\omega_c$: cavity frequency,
* $\omega_q$: qubit frequency,
* $g$: coupling strength.

If the **detuning** $\Delta = \omega_q - \omega_c \gg g$, then real energy exchange is suppressed.

Applying the SW transformation gives the **dispersive Hamiltonian**:

$$
H_{\text{eff}} \approx \omega_c a^\dagger a + \frac{\tilde{\omega}_q}{2}\sigma_z + \chi a^\dagger a \sigma_z
$$

where $\chi = g^2/\Delta$ is the **dispersive shift**.

So instead of real excitation swaps, the cavity frequency shifts depending on the qubit state (basis of **dispersive readout** in superconducting qubits).

---

### 🔹 Relevance in Quantum Computation

* Used to **derive effective Hamiltonians** when interactions are off-resonant.
* Essential for understanding **dispersive qubit–cavity readout**, **cross-Kerr interactions**, and **virtual ZZ couplings** in multi-qubit systems.
* Lets us design **gates mediated by virtual excitations**, without populating unwanted levels.

Awesome—let’s derive the **dispersive Hamiltonian** of circuit QED from the Jaynes–Cummings model using the **Schrieffer–Wolff (SW) transformation**.

---

# Setup

Start with the Jaynes–Cummings Hamiltonian (RWA) for a two-level qubit coupled to a single cavity mode:

$$
\frac{H}{\hbar}= \underbrace{\omega_c\,a^\dagger a + \frac{\omega_q}{2}\sigma_z}_{H_0}
\;+\;
\underbrace{g\,(a\,\sigma_+ + a^\dagger \sigma_-)}_{V},
$$

with detuning $\Delta=\omega_q-\omega_c$ and coupling $g$.
We assume the **dispersive regime**: $|\Delta|\gg g\sqrt{n+1}$ (no real swaps, only virtual).

Goal: find a unitary $e^{S}$ (with $S^\dagger=-S$) such that the transformed Hamiltonian $\tilde H=e^{S}He^{-S}$ has **no qubit–photon exchange terms to leading order**.

---

# Choose the generator $S$

Pick $S$ so that it cancels the off-diagonal coupling $V$ to first order:

$$
[S, H_0] = -V.
$$

A convenient choice is

$$
S=\frac{g}{\Delta}\,\big(a\,\sigma_+ - a^\dagger \sigma_- \big).
$$

You can check directly that $[S,H_0]=-\;g\,(a\,\sigma_+ + a^\dagger \sigma_-)= -V$.

---

# Expand the transformed Hamiltonian

Use the BCH expansion up to second order in $g/\Delta$:

$$
\tilde H
= H + [S,H] + \tfrac12 [S,[S,H]] + \cdots
= H_0 + V + [S,H_0] + [S,V] + \tfrac12 [S,[S,H_0]] + \cdots
$$

Since $[S,H_0]=-V$, the $V$ terms cancel:

$$
\tilde H = H_0 + \tfrac12 [S,V] + \mathcal{O}\!\left(\frac{g^3}{\Delta^2}\right).
$$

Compute the key commutator (using $\{\sigma_+,\sigma_-\}$ algebra: ${[\sigma_z, \sigma_+] = 2\sigma_+, [\sigma_z, \sigma_-] = -2 \sigma_-}$ and $[a,a^\dagger]=1$):

$$
[S,V] = \frac{g^2}{\Delta}\,\Big(2\,a^\dagger a\,\sigma_z + \sigma_z\Big).
$$

---

# Dispersive Hamiltonian (two-level qubit)

Putting it together:

$$
\frac{\tilde H}{\hbar}
= \omega_c a^\dagger a + \frac{\omega_q}{2}\sigma_z
\;+\; \frac{1}{2}\frac{g^2}{\Delta}\Big(2\,a^\dagger a\,\sigma_z + \sigma_z\Big)
$$

$$
= \Big(\omega_c + \chi\,\sigma_z\Big)a^\dagger a \;+\; \frac{1}{2}\Big(\omega_q + \chi\Big)\sigma_z,
\qquad
\chi \equiv \frac{g^2}{\Delta}.
$$

Common equivalent form:

$$
\frac{\tilde H}{\hbar} = \omega_c a^\dagger a + \frac{\omega_q}{2}\sigma_z + \chi\,a^\dagger a\,\sigma_z + \frac{\chi}{2}\sigma_z.
$$

**Interpretation**

* **Dispersive (cross-Kerr) shift**: cavity frequency shifts by $\pm \chi$ depending on the qubit state ($\sigma_z=\pm1$).
* **ac Stark/Lamb shift**: qubit frequency is shifted by $\chi(2n+1)$ where $n=\langle a^\dagger a\rangle$.
  (From $\chi a^\dagger a \sigma_z + \tfrac{\chi}{2}\sigma_z$.)

This is the foundation of **dispersive readout**: probe the cavity; its resonance moves by $\pm\chi$ depending on the qubit state.

---

# Extension to a transmon (weak anharmonicity $\alpha$)

A transmon isn’t perfectly two-level. Including the next level with anharmonicity $\alpha$ (negative for transmons), the effective dispersive shift becomes

$$
\chi \;\approx\; \frac{g^2\,\alpha}{\Delta(\Delta+\alpha)}.
$$

This captures the **level-repulsion from both $\lvert e\rangle\!\leftrightarrow\!\lvert g\rangle$** and **$\lvert f\rangle\!\leftrightarrow\!\lvert e\rangle$** transitions and is the standard formula used in experiments.

---

# Validity & notes

* **Regime**: $|\Delta|\gg g\sqrt{n+1}$ (and also $|\Delta+\alpha|\gg g\sqrt{n+1}$ for transmons).
* Keeps terms up to $\mathcal{O}(g^2/\Delta)$; higher orders give small Kerr terms like $K\,a^{\dagger 2}a^2$.
* The transformation produces an **effective, number-conserving Hamiltonian**: no real swaps, only **state-dependent shifts**.

---

# TL;DR

By choosing $S=\tfrac{g}{\Delta}(a\sigma_+ - a^\dagger\sigma_-)$ and expanding to second order, the SW transformation **integrates out** off-resonant qubit–photon exchange and yields the **dispersive Hamiltonian**:

$$
\boxed{\;\frac{\tilde H}{\hbar} = \omega_c a^\dagger a + \frac{\omega_q}{2}\sigma_z
+ \chi\,a^\dagger a\,\sigma_z + \frac{\chi}{2}\sigma_z,\quad \chi=\frac{g^2}{\Delta}\;}
$$

(or $\chi = g^2\alpha/[\Delta(\Delta+\alpha)]$ for a transmon), which underpins dispersive readout and ac-Stark shifts in circuit QED.


Great question 👍 — the **benefit of the Schrieffer–Wolff (SW) transformation** is that it gives us a **systematic way to eliminate unwanted couplings** and derive a simpler, effective Hamiltonian that still captures the physics of interest.

Here are the key advantages, especially in quantum systems like circuit QED:

---

### 🔑 Benefits of the SW transformation

1. **Removes fast oscillating/off-resonant terms**

   * In many problems (e.g., qubit–cavity coupling, Hubbard model with hopping, spin systems), you have small couplings to states that are far detuned in energy.
   * Directly including them makes the Hamiltonian messy and hard to solve.
   * SW transformation "rotates" into a frame where those couplings disappear to first order.

---

2. **Produces an effective Hamiltonian**

   * After the transformation, you get a Hamiltonian in which the qubit and cavity (in circuit QED) don’t exchange excitations directly, but instead you see **effective interactions** like dispersive shifts ($\chi a^\dagger a \sigma_z$).
   * This lets you understand the system in terms of measurable quantities: frequency shifts, Stark shifts, Kerr nonlinearities.

---

3. **Perturbative but systematic**

   * Instead of hand-waving away small terms, SW gives a **controlled expansion** in powers of $g/\Delta$ (coupling/detuning).
   * You know exactly the order of accuracy, and higher-order corrections (like photon–photon Kerr terms) can be derived if needed.

---

4. **Keeps Hermiticity & unitarity**

   * Unlike simply "dropping" terms (like in the Rotating Wave Approximation), the SW transformation applies a **unitary change of basis**, ensuring the effective Hamiltonian is still Hermitian and physical.

---

5. **Insight into virtual processes**

   * It shows explicitly how off-resonant couplings give rise to **virtual excitations** that shift energies (e.g., Lamb shift, Stark shift) without real excitations occurring.
   * That’s essential for understanding dispersive readout, cross-Kerr interactions, and error channels.

---

### 🧩 In circuit QED specifically

* It **justifies dispersive readout**: qubit state ↔ cavity frequency shift.
* It explains **state-dependent ac Stark shifts** that must be accounted for in gate calibration.
* It helps derive effective two-qubit interactions mediated by a shared cavity (cross-resonance, ZZ coupling).

---

✅ **In short:**
The Schrieffer–Wolff transformation is valuable because it turns a complicated Hamiltonian with small but messy couplings into a clean, effective description where the important low-energy physics (shifts, dispersive couplings) is transparent, controlled, and experimentally measurable.

Sweet — here’s a concrete **circuit QED** example that shows exactly why the **Schrieffer–Wolff (SW)** transformation is so useful compared with brute-force diagonalization.

---

# Problem setup (Jaynes–Cummings)

A qubit (frequency $\omega_q$) couples to a cavity (frequency $\omega_c$) with strength $g$:

$$
\frac{H}{\hbar}=\omega_c a^\dagger a + \frac{\omega_q}{2}\sigma_z + g\,(a\sigma_+ + a^\dagger\sigma_-).
$$

Detuning: $\Delta=\omega_q-\omega_c$.
**Dispersive regime:** $|\Delta|\gg g\sqrt{n+1}$.

---

# What SW gives you instantly

Choose $S=\tfrac{g}{\Delta}(a\sigma_+ - a^\dagger\sigma_-)$. Expanding $e^{S}He^{-S}$ to second order yields

$$
\frac{H_{\text{disp}}}{\hbar} = \omega_c a^\dagger a + \frac{\omega_q}{2}\sigma_z
\;+\; \chi\,a^\dagger a\,\sigma_z \;+\; \frac{\chi}{2}\sigma_z,
\qquad \chi=\frac{g^2}{\Delta}.
$$

### Immediate physical payoffs

* **Dispersive (cross-Kerr) shift**: the **cavity frequency** becomes $\omega_c \pm \chi$ depending on the qubit state ($\sigma_z=\pm1$).
  → Basis of **dispersive readout**: the cavity resonance moves by $2|\chi|$ between $|g\rangle$ and $|e\rangle$.
* **ac Stark/Lamb shift**: the **qubit frequency** becomes $\omega_q + \chi(2n+1)$ for $n$ photons in the cavity.
  → Lets you predict photon-number–dependent qubit shifts during readout drives.

For a **transmon** with anharmonicity $\alpha$ (negative), replace $\chi$ by

$$
\chi \approx \frac{g^2\,\alpha}{\Delta(\Delta+\alpha)},
$$

which matches experiments far better than a strict two-level model.

---

# Numbers you can use today

Pick realistic parameters (in Hz units, quoted as $x/2\pi$ for readability):

* $\omega_c/2\pi = 6\,\text{GHz}$,
* $\omega_q/2\pi = 5\,\text{GHz} \Rightarrow \Delta/2\pi = -1\,\text{GHz}$,
* $g/2\pi = 100\,\text{MHz}$.

Then

$$
\frac{\chi}{2\pi}=\frac{(g/2\pi)^2}{\Delta/2\pi}
= \frac{(100\,\text{MHz})^2}{-1\,\text{GHz}} = -10\,\text{MHz}.
$$

**Predictions from SW (no diagonalization needed):**

* Cavity resonance shifts by $\pm\chi \Rightarrow$
  $\omega_c^{(g)}/2\pi = 6.010\,\text{GHz}$,
  $\omega_c^{(e)}/2\pi = 5.990\,\text{GHz}$.
  Separation between responses = $2|\chi|/2\pi = 20\,\text{MHz}$.
* Qubit Stark shift at mean photon number $\bar n$:
  $\omega_q' \approx \omega_q + \chi(2\bar n + 1)$.
  With $\bar n=3$: shift $\approx -70\,\text{MHz}$.

You can design your readout drive and filter bandwidth just from these **back-of-the-envelope** numbers.

---

# What brute-force diagonalization would do

If you diagonalize the full Hamiltonian in the $n$-excitation manifolds, the dressed energies are

$$
\frac{E_{n,\pm}}{\hbar} = n\omega_c + \frac{\Delta}{2} \pm \sqrt{\left(\frac{\Delta}{2}\right)^2 + g^2 n}.
$$

To extract the **cavity pull** you must:

1. Build a truncated Hilbert space,
2. Diagonalize numerically for each $n$,
3. Take differences of dressed energies to infer effective frequencies,
4. Repeat vs. parameters to see trends.

In the dispersive limit you can Taylor-expand the square root and **recover the same $\chi$**, but SW gets you there **immediately**, cleanly, and with a controlled error $O(g^3/\Delta^2)$.

---

# Why SW is better for design & intuition

* **Single line formulae** for $\chi$, ac Stark, and (with higher order) Kerr terms → direct calibration targets.
* **Clarity:** you see the state-dependent cavity shift and photon-number–dependent qubit shift explicitly.
* **Scalable to multiqubit cases:** SW gives effective $ZZ$ couplings, cross-Kerr terms, etc., without huge matrices.

---

# Practical guardrails

* Valid if $|\Delta|\gg g\sqrt{n+1}$ (and $|\Delta+\alpha|\gg g\sqrt{n+1}$ for transmons).
* Watch the **critical photon number** $n_{\text{crit}} \sim \Delta^2/(4g^2)$; staying below it keeps you safely dispersive.
* Higher-order SW yields small cavity/self-Kerr terms useful for precise modeling.

---

## TL;DR

With the SW transformation you skip brute-force diagonalization and jump straight to a compact **effective Hamiltonian** whose parameters $(\chi, \text{Stark shifts}, ZZ\,\text{couplings})$ match what you measure and calibrate in the lab. It’s faster, clearer, and comes with a built-in accuracy knob (order in $g/\Delta$).


Below is a compact derivation of how a **transmon’s higher levels** correct the **dispersive shift** in circuit QED. The punchline is the well-known result

$$
\boxed{\;\chi \;=\; -\,\frac{g^2\,\alpha}{\Delta(\Delta+\alpha)}\;}
$$

where $g$ is the qubit–cavity coupling on the $|g\rangle\!\leftrightarrow\!|e\rangle$ transition, $\Delta=\omega_{ge}-\omega_c$ is the detuning, and $\alpha=\omega_{ef}-\omega_{ge}$ is the (negative) transmon anharmonicity. (For a transmon $\alpha<0$, so $\chi>0$ in practice.)

---

# Setup

Model the transmon as a weakly anharmonic oscillator (Duffing):

* Cavity: $H_c/\hbar=\omega_c a^\dagger a$.
* Transmon: levels $|m\rangle$ with $\omega_{m+1,m}=\omega_{ge}+m\alpha$.
* Interaction (RWA):

$$
H_{\text{int}}/\hbar
=\sum_{m\ge0} g_m\Big(a\,|m\!+\!1\rangle\langle m| + a^\dagger\,|m\rangle\langle m\!+\!1|\Big),
$$

with matrix elements $g_m \approx g\sqrt{m+1}$ (from the oscillator ladder matrix elements).

Define detunings for each adjacent transition:

$$
\Delta_m = \omega_{m+1,m}-\omega_c
= \Delta + m\alpha,\qquad \Delta\equiv \omega_{ge}-\omega_c .
$$

We work in the **dispersive regime**: $|\Delta_m|\gg g_m\sqrt{n+1}$ for relevant photon numbers $n$.

---

# Schrieffer–Wolff to second order

Choose an anti-Hermitian generator

$$
S=\sum_{m\ge0} \frac{g_m}{\Delta_m}\Big(a\,|m\!+\!1\rangle\langle m| - a^\dagger\,|m\rangle\langle m\!+\!1|\Big),
$$

which satisfies $[S,H_0]= - H_{\text{int}}$ for $H_0=H_c+H_{\text{q}}$.
Expanding $e^{S} H e^{-S}$ to second order gives

$$
H_{\text{eff}} \;=\; H_0 \;+\; \tfrac12 [S,H_{\text{int}}] \;+\; O(g^3/\Delta^2).
$$

The commutator generates **state-dependent dispersive shifts** of the cavity and qubit. One convenient way to collect terms is to write the cavity frequency conditioned on the transmon level $|m\rangle$ as

$$
\omega_c \;\to\; \omega_c + \chi_m,\qquad
\chi_m \;=\; \frac{g_m^2}{\Delta_m}\;-\;\frac{g_{m-1}^2}{\Delta_{m-1}},
$$

with the convention $g_{-1}=0$. (This is a standard result for a weakly anharmonic ladder under RWA.)

* For $|g\rangle$ ($m=0$): $\chi_g = \dfrac{g_0^2}{\Delta_0} = \dfrac{g^2}{\Delta}$.
* For $|e\rangle$ ($m=1$): $\chi_e = \dfrac{g_1^2}{\Delta_1}-\dfrac{g_0^2}{\Delta_0} = \dfrac{2g^2}{\Delta+\alpha}-\dfrac{g^2}{\Delta}$.

The **cavity pull** between qubit states (the coefficient of $a^\dagger a\,\sigma_z$) is

$$
2\chi \;=\; \chi_e - \chi_g
\;=\; \Big(\frac{2g^2}{\Delta+\alpha}-\frac{g^2}{\Delta}\Big) - \frac{g^2}{\Delta}
\;=\; -\,\frac{2g^2\,\alpha}{\Delta(\Delta+\alpha)}.
$$

Hence

$$
\boxed{\;\chi \;=\; -\,\frac{g^2\,\alpha}{\Delta(\Delta+\alpha)}\; }.
$$

Equivalently, you can write

$$
\chi \;=\; g^2\!\left(\frac{1}{\Delta}-\frac{1}{\Delta+\alpha}\right),
$$

which makes the **correction from the $e\!\leftrightarrow\!f$** transition explicit. In the **two-level limit** $|\alpha|\!\to\!\infty$ (i.e., no nearby higher level), this reduces to $\chi \to g^2/\Delta$. In the **harmonic limit** $\alpha\to 0$ the shift vanishes, as expected for a linear oscillator.

---

## Effective dispersive Hamiltonian (transmon)

Collecting terms, the second-order effective Hamiltonian is

$$
\frac{H_{\text{disp}}}{\hbar}
= \omega_c a^\dagger a
+ \frac{\tilde\omega_q}{2}\,\sigma_z
+ \chi\,a^\dagger a\,\sigma_z,
$$

with $\tilde\omega_q$ including the **Lamb/ac-Stark** renormalization (a constant piece $\propto \chi$) and
$\chi$ given above. Higher-order SW terms add small Kerr nonlinearities to both cavity and transmon.
