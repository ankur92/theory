# Decoherence and Control in the Harmonic Limit


## Introduction

As mentioned previously, for both capacitive and inductive driving, we can write the driving Hamiltonian in the lab frame as:

$$
H_d = h_d f(t) \sigma
$$

with:

|                | **Charge**                | **Flux**                 |
|----------------|---------------------------|--------------------------|
| $\sigma$      | $\sigma_y$                | $\sigma_x$              |
| $f(t)$        | $V_d(t)/V_d$              | $I_d(t)/I_d$            |
| $h_d$         | $Q_\text{zpf} V_d(C_d/C)$ | $-\Phi_\text{zpf} I_d (M_d/L)$ |

Using the forms of $Q_\text{zpf}$ and $\Phi_\text{zpf}$ given above, we can rewrite $h_d$ as:

$$
\text{(charge)} \quad \frac{h_d}{\hbar} = \frac{1}{\sqrt{2}} \left(\frac{C_d}{C} \right) \sqrt{\frac{(R_K / 8\pi)}{Z_q}} \frac{V_d}{\Phi_0/2\pi}
$$

$$
\text{(flux)} \quad \frac{h_d}{\hbar} = \frac{1}{\sqrt{2}} \left( \frac{M_d}{L} \right) \sqrt{\frac{Z_q}{(R_K/8\pi)}} \frac{I_d}{2e}
$$

where $Z_q$ is the qubit impedance and $R_K / 8 \pi = 1,027 \, \Omega$.

## Pi Pulse Length

For a resonant drive with envelope $f(t) = \cos(\omega t + \phi)$, we found in the rotating frame:

$$
H_d \, (\text{rotating frame}) = \frac{h_d}{2} \left( \cos(\phi) \sigma_x - \sin(\phi) \sigma_y \right).
$$

The angle $\phi$ is a phase reference for the drive signal, and we can choose it to be whatever we want. For convenience, we choose $\phi=0$ for the flux case and $\phi=\pi/2$ for the charge case, allowing us to write the unitary operator driving the qubit as:

$$
U(\Delta t) = \exp \left( -\frac{i}{\hbar} \int_0^{\Delta t} H_d(t') \, dt' \right) = \exp \left( -\frac{i}{\hbar} \frac{h_d}{2} \sigma \Delta t \right),
$$

where $\sigma$ stands for $\sigma_x$ for the flux case and $\sigma_y$ for the charge case. A $\pi$-pulse occurs when the argument of the exponent is equal to $\pm i \sigma \pi / 2$, i.e., when:

$$
\frac{1}{2} \frac{h_d}{\hbar} \Delta t = \pi / 2.
$$

Using the forms of $h_d/\hbar$ given above, we get the condition for a pi pulse as:

$$
\text{(charge)} \quad \frac{1}{\pi\sqrt{2}} \left( \frac{C_d}{C} \right) \sqrt{\frac{(R_K/8\pi)}{Z_q}} \frac{V_d \Delta t}{\Phi_0/2\pi} = 1,
$$

$$
\text{(flux)} \quad \frac{1}{\pi\sqrt{2}} \left( \frac{M_d}{L} \right) \sqrt{\frac{Z_q}{(R_K/8\pi)}} \frac{I_d \Delta t}{2e} = 1.
$$

Note that $V_d$ is just the amplitude of the sinusoidal drive voltage at the qubit's drive capacitor, and similarly for $I_d$.

## Decay from Spontaneous Emission

In the case of a nearly harmonic qubit like the transmon, the effect of dissipation from the drive line can be computed from classical electronics arguments. The quality factor $Q_d$ induced by the driving circuit is:

$$
Q_d (\text{capacitive}) = \left( \frac{C}{C_d} \right)^2 \frac{Z_q}{R_e},
$$

$$
Q_d (\text{inductive}) = \left( \frac{L}{M_d} \right)^2 \frac{R_e}{Z_q},
$$

where $R_e$ is the external line resistance, usually $50 \, \Omega$. Using these formulae, we can eliminate the capacitances and inductances from the formulae for the pi pulse length, arriving at:

$$
\text{(charge)} \quad \frac{1}{\pi\sqrt{2}} \sqrt{\frac{1}{Q_d}} \sqrt{\frac{(R_K/8\pi)}{R_e}} \frac{V_d \Delta t}{\Phi_0/2\pi} = 1,
$$

$$
\text{(flux)} \quad \frac{1}{\pi\sqrt{2}} \sqrt{\frac{1}{Q_d}} \sqrt{\frac{R_e}{(R_K/8\pi)}} \frac{I_d \Delta t}{2e} = 1.
$$

These equations link the loaded quality factor of the (harmonic) qubit to the external ($50 \, \Omega$) resistance, the pi-pulse length, and the driving amplitude.
